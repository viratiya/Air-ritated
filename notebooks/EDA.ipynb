# Air-ritated — Reproducible EDA
This notebook reads organizer files and generated audit artifacts. Run `python -m src.train` first to reproduce all model outputs.

```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
train = pd.read_csv(ROOT / 'data' / 'train.csv')
test = pd.read_csv(ROOT / 'data' / 'test.csv')
labels = pd.read_csv(ROOT / 'data' / 'test_labels.csv')
audit = json.loads((ROOT / 'outputs' / 'data_audit.json').read_text())
train.shape, test.shape, labels.shape
```

```python
audit['date_ranges'], audit['chronology'], audit['sentinel_minus_200']
```

```python
valid = train.loc[train['CO(GT)'] != -200].copy()
valid['DateTime'] = pd.to_datetime(valid['Date'] + ' ' + valid['Time'])
valid.set_index('DateTime')['CO(GT)'].resample('D').mean().plot(figsize=(12, 4), title='Daily mean CO(GT)')
plt.ylabel('CO (mg/m³)')
plt.show()
```
